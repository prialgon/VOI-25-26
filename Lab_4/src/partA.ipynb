```python
def show_image(img):
    cv2.imshow("Image", img)
    cv2.waitKey()
    cv2.destroyAllWindows()
    
def write_image(img, filename="img.jpg"):
    cv2.imwrite(filename, img)
    
def load_images(filenames):
    return [cv2.imread(filename) for filename in filenames]

base_path = os.path.join(os.getcwd(), "../data/partA/*g")

gb_path = glob.glob(base_path)
```

```python
names = ["figures", "football", "sudoku", "tennis"]
```

### **Tarea A.1.** Aplique `cv2.Canny()` a las imágenes de trabajo ajustando los hiperparámetros

```python
# TODO Apply canny to the images, play with the thresholds to get a good result

edge_imgs = load_images(gb_path)  # to store the canny edge images for later Hough transform

# Hyperparameters
low_threshold = 100
high_threshold = 200

os.makedirs("../data/partA/canny", exist_ok=True)

canny_edge_imgs = []
for i in range(len(edge_imgs)):
    # get the first image
    img = edge_imgs[i]

    # Convert to grayscale
    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Apply Canny edge detection
    edge_img = cv2.Canny(gray_img, low_threshold, high_threshold)

    # Store the edge image for later Hough transform
    canny_edge_imgs.append(edge_img)

    # Display the image or save it as you like
    write_image(edge_img, f"../data/partA/canny/{names[i]}_canny.png")
    
```

### **Tarea A.2.** Implemente `draw_lines()` para pintar las lineas sobre cada imagen

```python
# TODO Draw the lines on the images. A line is a tuple of 4 integers (x1, y1, x2, y2) where (x1, y1) and (x2, y2) are the coordinates of the two points of the line.
def draw_lines(img, lines, color, thickness):
    
    if len(lines) > 1:
        for line in lines:
            line = line.squeeze()
            cv2.line(img, (line[0], line[1]), (line[2], line[3]), color, thickness)
    else:
        lines = lines.squeeze()
        cv2.line(img, (lines[0], lines[1]), (lines[2], lines[3]), color, thickness)
        
    return img
```

### **Tarea A.3.** Aplique Hough con `cv2.HoughLinesP()` a las imágenes de trabajo para extraer líneas y afine los hiperparámetros

```python
# TODO Apply the hough transform to the images, play with the parameters to get a good result
from math import pi

# 1st image

# Define the hough transform parameters
rho = 1
theta = [pi/720, pi/180, pi/180, pi/720]
threshold = [30, 35, 60, 100]
min_line_length = [10, 50, 50, 50]
max_line_gap = [5, 20, 25, 50]

os.makedirs("../data/partA/hough", exist_ok=True)

hough_images = []
for i in range(len(edge_imgs)):
    if i != 3:
        continue
    # Get the edge image
    edges = canny_edge_imgs[i]

    # Apply the hough transform
    lines = cv2.HoughLinesP(edges, rho, theta[i], threshold[i], minLineLength=min_line_length[i], maxLineGap=max_line_gap[i])

    # Draw the lines on the image (on the original image, make sure to get a copy, as draw_lines modifies the image)
    line_img = draw_lines(edge_imgs[i].copy(), lines, (255, 0, 0), 2)

    # Show the image or save it as you like
    hough_images.append(line_img)
    
    show_image(line_img)
    
    write_image(line_img, f"../data/partA/hough/{names[i]}_hough.jpg")
```
